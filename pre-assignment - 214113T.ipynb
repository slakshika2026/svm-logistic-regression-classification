```python
# Importing Required Libraries
import pandas as pd
import numpy as np

from sklearn.model_selection import train_test_split
from sklearn.preprocessing import LabelEncoder, StandardScaler
from sklearn.linear_model import LogisticRegression
from sklearn.svm import SVC
from sklearn.metrics import accuracy_score
```

```python
# Loading the Dataset
df = pd.read_csv("credit_risk_synthetic.csv")
df.head()
```

```python
# Dataset Overview
df.info()
```

```python
# Check for missing values
print(df.isnull().sum())
```

```python
# Encoding Categorical Features

label_encoder = LabelEncoder()

df['home_ownership'] = label_encoder.fit_transform(df['home_ownership'])
df['purpose'] = label_encoder.fit_transform(df['purpose'])
```

```python
# Feature and Target Separation
X = df.drop('default', axis=1)
y = df['default']
```

```python
# Detect outliers using Z-score
z_scores = np.abs((X - X.mean()) / X.std())
outliers = np.where(z_scores > 3)

# Remove outliers
X_cleaned = X[(z_scores < 3).all(axis=1)]
y_cleaned = y[(z_scores < 3).all(axis=1)]

print("Number of data points after removing outliers:", len(X_cleaned))
```

```python
# Feature Scaling
scaler = StandardScaler()
X_scaled = scaler.fit_transform(X)
```

```python
# Train-Test Split
X_train, X_test, y_train, y_test = train_test_split(
    X_scaled, y, test_size=0.2, random_state=42
)
```

```python
# Logistic Regression Model Training
log_reg = LogisticRegression(max_iter=1000)
log_reg.fit(X_train, y_train)
```

```python
# Logistic Regression Evaluation
y_pred_log = log_reg.predict(X_test)
log_accuracy = accuracy_score(y_test, y_pred_log)

print("Logistic Regression Accuracy:", log_accuracy)
```

```python
# Support Vector Machine Model Training
svm_model = SVC(kernel='linear')
svm_model.fit(X_train, y_train)
```

```python
# Support Vector Machine Evaluation
y_pred_svm = svm_model.predict(X_test)
svm_accuracy = accuracy_score(y_test, y_pred_svm)

print("SVM Accuracy:", svm_accuracy)
```

```python
# Testing with a Random Data Sample
random_index = np.random.randint(0, X_test.shape[0])
sample = X_test[random_index].reshape(1, -1)

print("Actual Label:", y_test.iloc[random_index])
print("Logistic Regression Prediction:", log_reg.predict(sample)[0])
print("SVM Prediction:", svm_model.predict(sample)[0])
```

## Classification Results

- Logistic Regression achieved **98.75%** accuracy
- Support Vector Machine achieved **99.25%** accuracy

### Summary
The comparison shows that both classifiers perform effectively. However, the Support Vector Machine produces slightly higher accuracy for this classification task.
